```python
import pandas as pd

url = 'https://raw.githubusercontent.com/amdbraz/PI_G36/refs/heads/main/dataset_tratado.csv'
df = pd.read_csv(url)
```

```python
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
df['age'].plot(
    kind='hist',
    bins=10,
    color='#6baed6',
    edgecolor='white',
    alpha=0.8
)
plt.title('Distribuição de Idades', fontsize=12, fontweight='light')
plt.xlabel('Idade')
plt.ylabel('Frequência')
plt.box(False)
plt.show()
```

## Análise do gráfico de idade

```python
df.corr(numeric_only=True)
```

## Análise da matriz de correlação

Essa tabela mostra o quanto as variáveis numéricas do seu conjunto de dados estão relacionadas entre si. Os valores variam de -1 a 1:

Próximo de 1 → relação forte e positiva (quando uma aumenta, a outra também tende a aumentar).

Próximo de -1 → relação forte e negativa (quando uma aumenta, a outra tende a diminuir).

Próximo de 0 → relação fraca ou inexistente.

Relações fortes

current_weight_(lbs) e final_weight_(lbs) → 0.97  
Relação muito forte e positiva: quanto maior o peso atual, maior tende a ser o peso final.

current_weight_(lbs) e bmr_(calories) → 0.93  
Relação forte e positiva: pessoas com maior peso têm maior taxa metabólica basal (BMR).

bmr_(calories) e final_weight_(lbs) → 0.89  
Também forte e positiva: quanto maior o BMR, maior o peso final.

daily_calories_consumed e bmr_(calories) → 0.69  
Relação moderada a forte: quem consome mais calorias tende a ter maior BMR.

Relações fracas

age e bmr_(calories) → -0.19  
Relação fraca e negativa: o BMR tende a diminuir levemente com a idade.

stress_level e weight_change_(lbs) → -0.05  
Relação muito fraca: o nível de estresse não parece influenciar diretamente a variação de peso.

duration_(weeks) e quase todas as variáveis → valores próximos de 0
Indica que o tempo de duração do acompanhamento não tem correlação significativa com peso ou calorias.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calcula a matriz de correlação
corr = df.corr(numeric_only=True)

# Cria o heatmap com rótulos e escala de cores
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

plt.title("Matriz de Correlação entre Variáveis Numéricas")
plt.show()
```

## Análise do gráfico headmap

O gráfico mostra a correlação entre as variáveis numéricas do conjunto de dados.
As cores indicam a força e o sentido da relação:

Cores claras (próximo do amarelo/vermelho) → correlação positiva forte (quando uma variável aumenta, a outra também aumenta).

Cores escuras (azul/roxo) → correlação negativa (quando uma aumenta, a outra diminui).

Cores intermediárias → correlação fraca ou inexistente.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')

sns.boxplot(
    data=df,
    x='physical_activity_level',
    y='final_weight_(lbs)',
    palette=['#9ecae1', '#a1d99b', '#fdd0a2', '#bcbddc'],  # paleta suave
    width=0.6,
    linewidth=1
)

plt.title('Peso Final por Nível de Atividade Física', fontsize=12, fontweight='light', pad=15)
plt.xlabel('Nível de Atividade Física', fontsize=10)
plt.ylabel('Peso Final (lbs)', fontsize=10)
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## Análise do gráfico de peso final por nível de atividade fisica
O boxplot mostra como o peso final se distribui dentro de cada categoria de nível de atividade física.
Ele permite comparar rapidamente:

A mediana (linha central da caixa), indicando o peso mais típico de cada grupo.

A dispersão (tamanho da caixa), mostrando se os pesos variam pouco ou muito dentro do grupo.

Outliers, que aparecem como pontos isolados e representam valores fora do padrão.

Diferenças entre grupos, revelando se pessoas mais ativas tendem a ter pesos finais menores, maiores ou semelhantes às menos ativas.

```python
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')

df.groupby('physical_activity_level')['daily_calories_consumed'].mean().plot(
    kind='bar',
    color=['#9ecae1', '#a1d99b', '#fdd0a2', '#bcbddc'],
    edgecolor='white',
    alpha=0.9
)

plt.title('Média de Calorias por Nível de Atividade Física', fontsize=12, fontweight='light', pad=15)
plt.xlabel('Nível de Atividade Física', fontsize=10)
plt.ylabel('Calorias Médias Consumidas', fontsize=10)
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## Análise de média calorica por nível de atividade física
O gráfico compara o consumo diário de calorias entre os diferentes níveis de atividade física. Ele mostra como cada grupo tende a se alimentar, permitindo identificar padrões como:

Diferenças na média de calorias entre os níveis de atividade.

Se pessoas mais ativas consomem mais, menos ou quantidades semelhantes de calorias.

A relação entre gasto energético esperado e ingestão calórica.
